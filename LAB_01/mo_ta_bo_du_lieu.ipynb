# LAB 01 – Báo cáo mô tả bộ dữ liệu bài báo khoa học

**Nguồn:** OpenAlex API  
**Cách thu thập:** Tìm kiếm theo 6 nhóm từ khóa, lấy tối đa 25 bài/nhóm.
**Mục đích:** Giới thiệu sơ bộ bộ dữ liệu và thống kê mô tả cơ bản.

```python
import pandas as pd
from pathlib import Path

base_dir = Path.cwd()
if not (base_dir / "data" / "raw" / "research_papers_raw.csv").exists():
    base_dir = base_dir.parent
path = base_dir / "data" / "raw" / "research_papers_raw.csv"
df = pd.read_csv(path)
print("Số dòng:", df.shape[0], "| Số cột:", df.shape[1])
display(df.head())
```

## 1. Mô tả thuộc tính

| Thuộc tính | Ý nghĩa |
|---|---|
| paper_id | Mã OpenAlex của bài báo |
| title | Tiêu đề |
| query_field | Nhóm từ khóa được dùng để tìm bài |
| field | Lĩnh vực chính theo OpenAlex |
| authors | Danh sách tác giả |
| pub_year | Năm công bố |
| venue | Tạp chí/hội nghị/nguồn công bố |
| publisher | Tên tổ chức xuất bản (nếu có) |
| publication_type | Loại công trình |
| citations | Số lượt trích dẫn theo OpenAlex tại thời điểm thu thập |
| abstract | Tóm tắt |
| doi | Mã DOI |
| language | Ngôn ngữ |
| open_access | Có truy cập mở hay không |
| scholar_search_url | URL tìm theo tiêu đề trên Google Scholar |
| collected_at | Thời điểm thu thập |
| source_database | Cơ sở dữ liệu nguồn |
| source_url | URL bản ghi OpenAlex |

```python
print("\nThông tin cấu trúc:")
df.info()
print("\nSố giá trị thiếu:")
display(df.isna().sum().to_frame("missing_count"))
print("\nSố dòng trùng hoàn toàn:", df.duplicated().sum())
print("Số paper_id trùng:", df["paper_id"].duplicated().sum())
```

```python
print("Số bài theo nhóm truy vấn:")
display(df["query_field"].value_counts().to_frame("so_bai"))
print("Số bài theo năm:")
display(df["pub_year"].value_counts().sort_index().to_frame("so_bai"))
print("Thống kê citations:")
display(df["citations"].describe())
```
